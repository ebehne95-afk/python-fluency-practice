```python
budget = 10000
actual_cost = 12000
project_name = "Office renovation"

variance = actual_cost - budget
print(variance)
```

```python
if actual_cost > budget:
    status = "Over budget"

elif actual_cost == budget:
    status = "On budget"

else:
    status = "Under budget"

print(project_name, ":", status)
```
