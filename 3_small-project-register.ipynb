```python

projects = [
    {"name" : "Office renovation", "budget" : 10000, "actual_cost" : 12000},
    {"name" : "Lobby painting", "budget" : 3000, "actual_cost" : 2800},
    {"name" : "Roof repair", "budget" : 8000, "actual_cost" : 8000},
    {"name" : "Lighting replacement", "budget" : 4500, "actual_cost" : 4900},
    {"name" : "Flooring replacement", "budget" : 6000, "actual_cost" : 5500}
    ]

print(projects)
```

```python
total_budget = 0

for project in projects:
    variance = project["actual_cost"] - project["budget"]
    
    total_budget = total_budget + project["budget"]
    
    if variance > 0:
        status = "Over budget"
    elif variance == 0:
        status = "On budget"
    else:
        status = "Under budget"
    
    print(project["name"],
        project["budget"],
        project["actual_cost"],
        variance, status)
    
print(total_budget)
```

Above means take each dictionary in projects, call it project, and print its name.

```python
# Add a sixth project without copying report logic

projects.append({"name" : "Drywall repair", "budget" : 20000, "actual_cost" : 30000})
```

```python
total_budget = 0
total_actual_cost = 0
over_budget_count = 0
on_budget_count = 0
under_budget_count = 0

for project in projects:
    variance = project["actual_cost"] - project["budget"]
    
    total_budget = total_budget + project["budget"]

    total_actual_cost = total_actual_cost + project["actual_cost"]
    
    if variance > 0:
        status = "Over budget"
        over_budget_count = over_budget_count + 1
    elif variance == 0:
        status = "On budget"
        on_budget_count = on_budget_count +1
    else:
        status = "Under budget"
        under_budget_count = under_budget_count + 1
    
    print(project["name"],
        project["budget"],
        project["actual_cost"],
        variance, status)
    
print(total_budget, total_actual_cost)

total_variance = total_actual_cost - total_budget
print(total_variance, ",", status)

print(over_budget_count, on_budget_count, under_budget_count)
```
