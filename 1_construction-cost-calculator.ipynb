```python
quantity = 25
unit_price = 40.00
cont_rate = .10

print("Quantity:", quantity)
print("Unit price in dollars:", "$",unit_price)
print("Contingency rate:", ((cont_rate)*100),"%")
```

```python
base_cost = quantity * unit_price
print("Base cost (in dollars):", "$", base_cost)
```

```python
cont_amount = base_cost * cont_rate
print("Contingency amount:", "$", cont_amount)
```

```python
total = cont_amount + base_cost
print("Total:", "$", total)
```

Changing quantity to predict total cost. If I change the quantity from 25 to 30, the base cost will be $1200, the contingency amount will be $120, and the total will be $1320.

```python
quantity = 30
base_cost2 = quantity * unit_price
cont_amount2 = base_cost2 * cont_rate
total = cont_amount2 + base_cost2

print("Base cost:", "$", base_cost2)
print("Contingency amount:", "$", cont_amount2)
print("Total:", "$", total)
```
