```python
quantity = 25
unit_price = 40.00
cont_rate = .10

print("Quantity:", quantity)
print("Unit price in dollars:", "$",unit_price)
print("Contingency rate:", ((cont_rate)*100),"%")
```

```python
def calculate_total_cost(quantity, unit_price):
    return quantity * unit_price

total_cost = calculate_total_cost(25, 40)
print(total_cost)
```

```python
def calculate_variance(budget, actual_cost):
    return actual_cost - budget

variance = calculate_variance(10000, 12000)
print(variance)
```

`return` sends a result back to the caller so you can store or reuse it.

`print` shows a message; it does not return the displayed number for reuse.

```python
def calculate_pert_variance(variance, budget):
    return (variance / budget) * 100

pert_variance = calculate_pert_variance(2000, 10000)

print(pert_variance)
```

```python
pert_variance = calculate_pert_variance(variance, 10000)
print(pert_variance)
```

```python
def calculate_pert_variance(variance, budget):
    if budget == 0:
        return None
    return (variance / budget) * 100

pert_variance = calculate_pert_variance(2000, 10000)

print(pert_variance)

print(calculate_pert_variance(2000, 0))
```

What the above does is avoid dividing by zero. Next, I want to try two calls myself. A variance of -2,000 and a variance of 0, both with a budget 10,000. Predict percentages.

Negative variance means the project is under budget. A variance of 0 means the project is at budget.

The at budget project will not have a percentage of variance.

The under budget project will have a percentage of variance -20.

```python
pert_variance_a = calculate_pert_variance(-2000, 10000)
print(pert_variance_a)

pert_variance_b = calculate_pert_variance(0, 10000)
print(pert_variance_b)
```

I was mostly right. A 0% variance is a defined result. A budget of 0 with a 2000 variance is an undefined result (dividing byb 0); that will trigger a `None` response.
